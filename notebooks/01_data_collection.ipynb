```python
import numpy as np

def haversine_distance(lat1, lon1, lat2, lon2):
    """
    Calculate distance between two GPS coordinates.
    Returns distance in meters.
    """
    R = 6371000  # Earth radius in meters

    lat1 = np.radians(lat1)
    lat2 = np.radians(lat2)
    dlat = np.radians(lat2 - lat1)
    dlon = np.radians(lon2 - lon1)

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))

    return R * c
```

```python
import pandas as pd
import numpy as np

def process_trajectory(file_path):
    """
    Read and preprocess one GeoLife .plt trajectory file.
    """

    # Read raw trajectory
    data = pd.read_csv(
        file_path,
        skiprows=6,
        header=None
    )

    # Correct GeoLife columns
    data.columns = [
        "latitude",
        "longitude",
        "altitude",
        "unused",
        "days",
        "date",
        "time"
    ]

    # Create timestamp
    data["timestamp"] = pd.to_datetime(
        data["date"].astype(str) + " " + data["time"].astype(str),
        errors="coerce"
    )

    # Remove unnecessary columns
    data = data.drop(
        columns=["unused", "date", "time"]
    )

    # Remove invalid timestamps
    data = data.dropna(subset=["timestamp"])

    # Remove invalid GPS coordinates
    data = data[
        (data["latitude"].between(-90, 90)) &
        (data["longitude"].between(-180, 180))
    ]

    # Sort by time
    data = data.sort_values("timestamp").reset_index(drop=True)

    # Calculate distance
    data["distance_m"] = haversine_distance(
        data["latitude"].shift(1),
        data["longitude"].shift(1),
        data["latitude"],
        data["longitude"]
    ).fillna(0)

    # Calculate time difference
    data["time_diff_sec"] = (
        data["timestamp"].diff().dt.total_seconds()
    ).fillna(0)

    # Calculate speed
    data["speed_kmh"] = np.where(
        data["time_diff_sec"] > 0,
        (data["distance_m"] / data["time_diff_sec"]) * 3.6,
        0
    )

    # Flag unrealistic speed
    data["speed_outlier"] = data["speed_kmh"] > 120

    # Clean speed
    data["clean_speed_kmh"] = data["speed_kmh"].where(
        data["speed_kmh"] <= 120,
        np.nan
    )

    # Time features
    data["hour"] = data["timestamp"].dt.hour
    data["weekday"] = data["timestamp"].dt.day_name()
    data["is_weekend"] = (
        data["timestamp"].dt.dayofweek >= 5
    )

    # Movement indicator
    data["is_moving"] = data["clean_speed_kmh"] > 1

    return data
```

```python
from pathlib import Path
import pandas as pd
import time

# Find all trajectory files
geo_path = Path(
    "../data/raw/Geolife Trajectories 1.3/"
    "Geolife Trajectories 1.3/Data"
)

plt_files = list(geo_path.rglob("*.plt"))

print("Total trajectory files:", len(plt_files))

# Store processed data
all_trajectories = []

start_time = time.time()

for i, file_path in enumerate(plt_files, start=1):

    try:
        processed = process_trajectory(file_path)

        # Add user ID from folder name
        processed["user_id"] = file_path.parent.parent.name

        all_trajectories.append(processed)

    except Exception as e:
        print(f"Error processing {file_path}: {e}")

    # Show progress every 500 files
    if i % 500 == 0:
        elapsed = time.time() - start_time
        print(f"Processed {i}/{len(plt_files)} files "
              f"({elapsed:.1f} seconds)")

print("\nProcessing complete!")

# Combine everything
all_data = pd.concat(
    all_trajectories,
    ignore_index=True
)

print("Total GPS records:", len(all_data))
print("Total columns:", len(all_data.columns))
```

```python
# ============================================================
# STEP 65: ANALYZE LARGEST PREDICTION ERRORS
# ============================================================

import numpy as np
import pandas as pd

print("Step 65: Analyzing largest prediction errors...")
print("=" * 70)

# ------------------------------------------------------------
# 1. Get actual coordinates
# ------------------------------------------------------------

actual_lat = y_test_sample["next_latitude"].to_numpy()
actual_lon = y_test_sample["next_longitude"].to_numpy()

# ------------------------------------------------------------
# 2. Get predicted coordinates
# ------------------------------------------------------------

predicted_lat = predictions[:, 0]
predicted_lon = predictions[:, 1]

# ------------------------------------------------------------
# 3. Calculate geographic error using Haversine formula
# ------------------------------------------------------------

def haversine_distance(lat1, lon1, lat2, lon2):

    R = 6371.0  # Earth radius in km

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    c = 2 * np.arcsin(np.sqrt(a))

    return R * c * 1000


error_m = haversine_distance(
    actual_lat,
    actual_lon,
    predicted_lat,
    predicted_lon
)

print("Prediction errors calculated!")

print("\nMean error:",
      np.mean(error_m), "meters")

print("Median error:",
      np.median(error_m), "meters")

print("Minimum error:",
      np.min(error_m), "meters")

print("Maximum error:",
      np.max(error_m), "meters")


# ------------------------------------------------------------
# 4. Create error dataframe
# ------------------------------------------------------------

error_analysis = pd.DataFrame({

    "user_id": user_ids[test_indices],

    "actual_latitude": actual_lat,
    "actual_longitude": actual_lon,

    "predicted_latitude": predicted_lat,
    "predicted_longitude": predicted_lon,

    "error_m": error_m
})


# ------------------------------------------------------------
# 5. Sort by largest error
# ------------------------------------------------------------

largest_errors = (
    error_analysis
    .sort_values("error_m", ascending=False)
    .head(20)
)


print("\n" + "=" * 70)
print("TOP 20 LARGEST PREDICTION ERRORS")
print("=" * 70)

print(
    largest_errors.to_string(index=False)
)


# ------------------------------------------------------------
# 6. Count extreme errors
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ERROR SUMMARY")
print("=" * 70)

print(
    "Errors > 1 km:",
    np.sum(error_m > 1000)
)

print(
    "Errors > 10 km:",
    np.sum(error_m > 10000)
)

print(
    "Errors > 50 km:",
    np.sum(error_m > 50000)
)

print(
    "Errors > 100 km:",
    np.sum(error_m > 100000)
)


# ------------------------------------------------------------
# 7. Save results
# ------------------------------------------------------------

output_path = "../data/step65_largest_errors.csv"

error_analysis.to_csv(
    output_path,
    index=False
)

print("\nResults saved to:")
print(output_path)

print("\nStep 65 completed successfully!")
```

```python
# ============================================================
# SINGLE CELL: CREATE LIGHTWEIGHT DEPLOYMENT MODEL
# ============================================================

import os
import joblib
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor

print("=" * 70)
print("CREATING LIGHTWEIGHT DEPLOYMENT MODEL")
print("=" * 70)


# ------------------------------------------------------------
# 1. Locate and load processed GPS dataset
# ------------------------------------------------------------

print("\n[1/7] Loading processed GPS dataset...")

possible_paths = [
    "../data/processed/geolife_processed.parquet",
    "data/processed/geolife_processed.parquet",
    "../data/processed/processed_gps.parquet",
    "data/processed/processed_gps.parquet"
]

dataset_path = None

for path in possible_paths:
    if os.path.exists(path):
        dataset_path = path
        break

if dataset_path is None:
    raise FileNotFoundError(
        "geolife_processed.parquet could not be found.\n"
        "Expected location:\n"
        "data/processed/geolife_processed.parquet"
    )

print("Dataset found:")
print(os.path.abspath(dataset_path))

df = pd.read_parquet(dataset_path)

print("Dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))


# ------------------------------------------------------------
# 2. Recreate ML dataset
# ------------------------------------------------------------

print("\n[2/7] Recreating ML dataset...")

df = df.sort_values(
    ["user_id", "timestamp"]
).reset_index(drop=True)

# Previous location
df["previous_latitude"] = (
    df.groupby("user_id")["latitude"].shift(1)
)

df["previous_longitude"] = (
    df.groupby("user_id")["longitude"].shift(1)
)

# Next location
df["next_latitude"] = (
    df.groupby("user_id")["latitude"].shift(-1)
)

df["next_longitude"] = (
    df.groupby("user_id")["longitude"].shift(-1)
)

# Remove rows without previous/next location
ml_data = df.dropna(
    subset=[
        "previous_latitude",
        "previous_longitude",
        "next_latitude",
        "next_longitude"
    ]
).reset_index(drop=True)

print("ML dataset recreated!")
print("ML rows:", len(ml_data))


# ------------------------------------------------------------
# 3. Create the same 80/20 user-based split
# ------------------------------------------------------------

print("\n[3/7] Creating user-based train/test split...")

unique_users = np.array(
    ml_data["user_id"].unique()
)

print("Total users:", len(unique_users))

train_users, test_users = train_test_split(
    unique_users,
    test_size=0.20,
    random_state=42
)

print("Training users:", len(train_users))
print("Testing users :", len(test_users))


# ------------------------------------------------------------
# 4. Prepare features and targets
# ------------------------------------------------------------

print("\n[4/7] Preparing features...")

features = [
    "latitude",
    "longitude",
    "previous_latitude",
    "previous_longitude",
    "altitude",
    "clean_speed_kmh",
    "distance_m",
    "time_diff_sec",
    "hour",
    "is_weekend"
]

targets = [
    "next_latitude",
    "next_longitude"
]

X = ml_data[features].copy()
y = ml_data[targets].copy()

# Replace infinity values
X = X.replace(
    [np.inf, -np.inf],
    np.nan
)

# Fill missing numeric values
for col in X.columns:
    if X[col].isna().any():
        X[col] = X[col].fillna(
            X[col].median()
        )

print("Features:", len(features))
print("Targets :", len(targets))


# ------------------------------------------------------------
# 5. Select training users
# ------------------------------------------------------------

print("\n[5/7] Selecting training data...")

user_ids = ml_data["user_id"].reset_index(drop=True)

train_mask = user_ids.isin(train_users)

X_train = X.loc[train_mask]
y_train = y.loc[train_mask]

print("Full training rows:", len(X_train))


# ------------------------------------------------------------
# 6. Create SMALL deployment training sample
# ------------------------------------------------------------

print("\n[6/7] Creating lightweight training sample...")

# IMPORTANT:
# 100,000 rows instead of 500,000+
# 30 trees instead of 100
# max_depth=20 limits model size

SAMPLE_SIZE = min(
    100000,
    len(X_train)
)

sample_indices = X_train.sample(
    n=SAMPLE_SIZE,
    random_state=42
).index

X_sample = X_train.loc[
    sample_indices
]

y_sample = y_train.loc[
    sample_indices
]

print("Deployment training sample:")
print("X:", X_sample.shape)
print("y:", y_sample.shape)


# ------------------------------------------------------------
# 7. Train and save lightweight model
# ------------------------------------------------------------

print("\nTraining lightweight Random Forest...")
print("Please wait...")

deployment_model = RandomForestRegressor(
    n_estimators=30,
    max_depth=20,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

deployment_model.fit(
    X_sample,
    y_sample
)

print("Training completed successfully!")


# ------------------------------------------------------------
# Save model
# ------------------------------------------------------------

output_directory = "../data/final_model"

# If notebook is running from project root
if not os.path.exists(output_directory):
    output_directory = "data/final_model"

os.makedirs(
    output_directory,
    exist_ok=True
)

model_path = os.path.join(
    output_directory,
    "gps_location_prediction_deployment_model.pkl"
)

print("\nSaving deployment model...")

joblib.dump(
    deployment_model,
    model_path,
    compress=3
)

# ------------------------------------------------------------
# Model size
# ------------------------------------------------------------

model_size_mb = (
    os.path.getsize(model_path)
    / (1024 * 1024)
)

model_size_gb = model_size_mb / 1024

print("\n" + "=" * 70)
print("DEPLOYMENT MODEL CREATED SUCCESSFULLY")
print("=" * 70)

print("\nModel path:")
print(os.path.abspath(model_path))

print("\nModel size:")
print(f"{model_size_mb:.2f} MB")
print(f"{model_size_gb:.2f} GB")

print("\nModel configuration:")
print("Trees      :", 30)
print("Max depth  :", 20)
print("Training rows:", SAMPLE_SIZE)

print("\nUser split:")
print("Total users   :", len(unique_users))
print("Training users:", len(train_users))
print("Testing users :", len(test_users))

print("\n" + "=" * 70)
print("STEP COMPLETED")
print("=" * 70)
```
